```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Corrected path to the screenshot image actually present in your files
image_path = 'Screenshot 2026-10-15 135017.png'

img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if img is None:
    print(f"Error: Could not load image from '{image_path}'")
else:
    kernel = np.ones((5, 5), np.uint8)
    dilated_img = cv2.dilate(img, kernel, iterations=1)
    cv2_imshow(img)
    cv2_imshow(dilated_img)
    cv2.imwrite('Dilated_Image.jpg', dilated_img)
```

```python
import cv2
from google.colab.patches import cv2_imshow

# Corrected the image path to use the available screenshot image
img = cv2.imread('Screenshot 2026-10-15 135017.png')

if img is None:
    print("Error: Could not load the image. Please verify the filename.")
else:
    gray_image = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))
    eroded_image = cv2.erode(gray_image, kernel, iterations=1)
    cv2_imshow(gray_image)
    cv2_imshow(eroded_image)
    cv2.imwrite('eroded_image.jpg', eroded_image)
```

```python
def change_speed(src, dst, speed):
    cap = cv2.VideoCapture(src)
    fps = cap.get(cv2.CAP_PROP_FPS)
    w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    out = cv2.VideoWriter(dst, cv2.VideoWriter_fourcc(*'mp4v'), fps * speed, (w, h))
    count = 0
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        out.write(frame)
        count += 1
    cap.release()
    out.release()
    print(dst, "created. Frames:", count)

change_speed('video.mp4', 'slow_motion.mp4', 0.5)   # half speed
change_speed('video.mp4', 'fast_motion.mp4', 2.0)   # double speed

# convert so the browser can play them
!ffmpeg -y -i slow_motion.mp4 -vcodec libx264 slow_h264.mp4 -loglevel error
!ffmpeg -y -i fast_motion.mp4 -vcodec libx264 fast_h264.mp4 -loglevel error

from IPython.display import Video, display
print("SLOW MOTION:")
display(Video('slow_h264.mp4', embed=True, width=400))
print("FAST MOTION:")
display(Video('fast_h264.mp4', embed=True, width=400))
```

```python
import cv2
from google.colab.patches import cv2_imshow

# Using the image file that exists in your files
image_path = 'Screenshot 2026-10-15 135017.png'
image = cv2.imread(image_path)

if image is None:
    print(f"Error: Could not load image from '{image_path}'")
else:
    height, width = image.shape[:2]

    bigger_image = cv2.resize(image, (int(width * 3.0), int(height * 3.0)))
    smaller_image = cv2.resize(image, (int(width * 0.5), int(height * 0.5)))

    cv2_imshow(image)
    cv2_imshow(bigger_image)
    cv2_imshow(smaller_image)
    cv2.imwrite('Bigger_image.jpg', bigger_image)
    cv2.imwrite('smaller_image.jpg', smaller_image)
```
